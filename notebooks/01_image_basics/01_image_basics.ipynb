```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# A display helper that works in both Colab and locally
def show(*images, titles=None, cmap=None):
    titles = titles or [""] * len(images)
    plt.figure(figsize=(5 * len(images), 5))
    for i, img in enumerate(images):
        plt.subplot(1, len(images), i + 1)
        if img.ndim == 2:
            plt.imshow(img, cmap=cmap or "gray")
        else:
            plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
        plt.title(titles[i])
        plt.axis("off")
    plt.tight_layout()
    plt.show()

print("OpenCV version:", cv2.__version__)
```

## 2. Guided example - loading and inspecting an image

An image is just a NumPy array. OpenCV loads it in **BGR** order (not RGB), which is a common source of confusion.

```python
img = cv2.imread("resources/images/hasan.jpg")

print("Type        :", type(img))
print("Shape (H,W,C):", img.shape)
print("Data type   :", img.dtype)
print("Min / Max   :", img.min(), "/", img.max())

show(img, titles=["Original image"])
```

## 3. Guided example - colour channels

Splitting the channels lets us see how much red, green and blue each pixel contains. Remember OpenCV order is B, G, R.

```python
b, g, r = cv2.split(img)

show(b, g, r, titles=["Blue channel", "Green channel", "Red channel"])
```

## 4. Guided example - crop and resize

```python
h, w = img.shape[:2]

# Crop the central half of the image
crop = img[h // 4 : 3 * h // 4, w // 4 : 3 * w // 4]

# Resize to a fixed width
resized = cv2.resize(img, (200, 200))

show(crop, resized, titles=["Centre crop", "Resized 200x200"])
```

## 10. Interactive exploration - resolution and pixelation

Drag the slider to **downsample** the image and then **upscale** it back with nearest-neighbour. This shows what "resolution" really means: how many samples we kept.

```python
import ipywidgets as widgets
from ipywidgets import interact

def resize_demo(scale=100):
    h, w = img.shape[:2]
    small = cv2.resize(img, (max(1, w * scale // 100), max(1, h * scale // 100)),
                       interpolation=cv2.INTER_NEAREST)
    big = cv2.resize(small, (w, h), interpolation=cv2.INTER_NEAREST)
    show(big, titles=[f"kept {scale}% of pixels, then upscaled (pixelation)"])

interact(resize_demo, scale=widgets.IntSlider(min=5, max=100, step=5, value=100))
```

## 11. Check your understanding (Q&A)

<details><summary><b>Q1. What does the third dimension of <code>img.shape</code> represent, and why is it 3?</b></summary>

It is the number of colour channels. A standard colour image has three (Blue, Green, Red in OpenCV order). Grayscale images have no third dimension.

</details>

<details><summary><b>Q2. Why does OpenCV load images as BGR, and how do you fix the display in Matplotlib?</b></summary>

OpenCV's historic default is BGR. Matplotlib expects RGB, so convert with <code>cv2.cvtColor(img, cv2.COLOR_BGR2RGB)</code>; otherwise red and blue swap.

</details>

<details><summary><b>Q3. What is the difference between resizing and cropping?</b></summary>

Cropping selects a region (same resolution, fewer pixels); resizing changes the pixel grid and therefore the resolution.

</details>
